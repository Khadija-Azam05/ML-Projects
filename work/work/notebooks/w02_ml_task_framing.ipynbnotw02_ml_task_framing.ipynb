## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

## My lane as an ML task (type)

**Task Type:** Ranking

I am framing my lane as a **ranking** task because the goal is to prioritize content pages that should be reviewed first. Instead of predicting only "review" or "do not review," the model ranks content based on search performance signals so the SEO or content team can focus on the highest-priority pages first.

```python
print("ML Task Type: Ranking")
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

## Target or Proxy

The dataset does not contain a direct label indicating whether content should be refreshed. Therefore, I would use a **proxy target** based on observed search performance. Pages or content-query pairs with low clicks, declining visibility, poor average position, or low impression share could be assigned a higher refresh priority. This proxy is derived from observed data rather than a manually labeled outcome.

```python
df[[
    "impressions_90d",
    "clicks_90d",
    "avg_position_90d",
    "content_total_impressions_90d"
]].head()
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

## Success Metric

The success of this task will be measured by how well the model ranks high-priority content near the top of the list. A good model helps the content team quickly identify which pages should be reviewed first, reducing the time spent reviewing lower-priority pages. The ranking should consistently place the most valuable refresh opportunities near the top.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML beats a fixed rule here

A fixed rule cannot capture the complex relationships between search performance metrics. For example, a page with low clicks may still perform well if it has low search volume, while another page with similar clicks but many impressions may need attention. Machine learning can learn patterns from multiple features, such as impressions, clicks, average position, query visibility, and content performance, to produce a better ranking than a single threshold-based rule.
